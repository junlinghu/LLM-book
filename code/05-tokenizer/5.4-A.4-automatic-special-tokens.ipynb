# A.4: Automatic special tokens

BERT's tokenizer adds `[CLS]` and `[SEP]` by itself.

Earlier imports and definitions this listing relies on are included below so the notebook runs on its own.


````python
# Imports and definitions from earlier listings, included so this notebook runs alone.

# From A.2 Greedy longest-match encoding
def wordpiece_encode(word, vocab, unk="[UNK]"):
    """Greedy longest-match-first segmentation of a single word."""
    pieces, start = [], 0
    while start < len(word):
        end = len(word)
        while end > start:
            piece = word[start:end] if start == 0 else "##" + word[start:end]
            if piece in vocab:
                break
            end -= 1
        if end == start:              # no piece matches: the whole word is unknown
            return [unk]
        pieces.append(piece)
        start = end
    return pieces

# From A.2 Greedy longest-match encoding
from transformers import AutoTokenizer

bert = AutoTokenizer.from_pretrained("bert-base-uncased")

vocab = bert.get_vocab()

for w in ["tokenization", "unaffable", "playing", "xyzzy"]:
    print(w, wordpiece_encode(w, vocab), bert.tokenize(w))
````
