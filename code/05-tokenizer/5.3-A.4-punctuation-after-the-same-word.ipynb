# A.4: Punctuation after the same word

Show that " dog" is the same token before different punctuation marks.


````python
import tiktoken

enc = tiktoken.get_encoding("gpt2")
for s in [" dog.", " dog!", " dog?"]:
    ids = enc.encode(s)
    print(repr(s), ids, [enc.decode([i]) for i in ids])
````

**Expected output**

Output:

```text
' dog.' [3290, 13] [' dog', '.']
' dog!' [3290, 0] [' dog', '!']
' dog?' [3290, 30] [' dog', '?']
```
