# 6.2: Embedding lookup with √d scaling


```python
import math
import torch
import torch.nn as nn

V, d = 1000, 512
emb = nn.Embedding(V, d)
tokens = torch.tensor([[5, 17, 42, 3]])      # a batch with one sequence of 4 token IDs
X = emb(tokens) * math.sqrt(d)               # shape (1, 4, 512)
print(X.shape)
```
