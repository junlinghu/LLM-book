# Code 2.3.3: The hand-built XOR network

Evaluates the ReLU solution to XOR on all four inputs and prints the hidden representation and the output. It reuses `np` from Code 2.3.1.

Earlier imports and definitions this listing relies on are included below so the notebook runs on its own.


```python
# Imports and definitions from earlier listings, included so this notebook runs alone.
# From Code 2.3.1
import numpy as np
```

```python
relu = lambda z: np.maximum(0, z)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
W1 = np.array([[1.0, 1.0],
               [1.0, 1.0]])
b1 = np.array([0.0, -1.0])
W2 = np.array([1.0, -2.0])
H = relu(X @ W1 + b1)
print("hidden:", H.tolist(), " output:", H @ W2)
# hidden: [[0.0, 0.0], [1.0, 0.0], [1.0, 0.0], [2.0, 1.0]]  output: [0. 1. 1. 0.]
```
