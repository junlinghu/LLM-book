# A.3: Whitespace and capitalization

Variants of "hello" and an indented line of code.

Earlier imports and definitions this listing relies on are included below so the notebook runs on its own.


````python
# Imports and definitions from earlier listings, included so this notebook runs alone.

# From A.1 Setup and spelling
import tiktoken

encs = {n: tiktoken.get_encoding(n) for n in ["gpt2", "cl100k_base", "o200k_base"]}

enc = encs["gpt2"]

def show(e, s):
    return [e.decode([i]) for i in e.encode(s)]
````

````python
for w in ["hello", " hello", "Hello", " Hello", " HELLO"]:
    print(repr(w), enc.encode(w))
````

**Expected output**

Output:

```text
'hello' [31373]
' hello' [23748]
'Hello' [15496]
' Hello' [18435]
' HELLO' [47899, 46]
```


````python
line = "        return x"
print(show(enc, line))
print(show(encs["cl100k_base"], line))
````

**Expected output**

Output:

```text
[' ', ' ', ' ', ' ', ' ', ' ', ' ', ' return', ' x']
['       ', ' return', ' x']
```
