# Code 2.3.2: Checking the 2-2-1 worked example

Reproduces the hand calculation of the worked example: pre-activations, tanh activations, logit, output probability, and cross-entropy loss. It reuses `np` from Code 2.3.1.

Earlier imports and definitions this listing relies on are included below so the notebook runs on its own.


```python
# Imports and definitions from earlier listings, included so this notebook runs alone.
# From Code 2.3.1
import numpy as np
```

```python
x = np.array([1.0, 0.5])
W1 = np.array([[0.5, -0.3],
               [0.8,  0.2]])
b1 = np.array([0.0, 0.1])
W2 = np.array([1.0, -1.5])
b2 = 0.2

z1 = x @ W1 + b1
h = np.tanh(z1)
z2 = h @ W2 + b2
p = 1 / (1 + np.exp(-z2))
print("z1 =", z1, " h =", h.round(4), f" z2 = {z2:.4f}  p = {p:.4f}  loss = {-np.log(p):.4f}")
# z1 = [ 0.9 -0.1]  h = [ 0.7163 -0.0997]  z2 = 1.0658  p = 0.7438  loss = 0.2960
```
